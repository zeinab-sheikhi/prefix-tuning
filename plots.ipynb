```python
import matplotlib.pyplot as plt
# Sample data (replace these values with your actual data)
prefix_lengths = [5, 10, 15]
lr5_scores = [13.414, 12.3649, 12.26 ]
lr8_scores = [10.8916, 10.8003, 12.77 ]


# Create a figure and axis objects
fig, ax1 = plt.subplots()



# Add a y-axis for lr=8e-05
ax2 = ax1.twinx()
ax2.plot(prefix_lengths, lr8_scores, 'o-', color='red', label='BLEU, lr=8e-05')
ax2.set_ylabel('BLEU, lr=8e-05', color='red')
ax2.tick_params(axis='y', labelcolor='red')

# Add a y-axis for lr=5e-05
ax1.plot(prefix_lengths, lr5_scores, 'o-', color='blue', label='BLEU, lr=5e-05')
ax1.set_xlabel('Prefix Length')
ax1.set_ylabel('BLEU, lr=5e-05', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')
# Add legend
fig.legend(loc='center')

# Show plot
plt.tight_layout()
plt.show()
```
